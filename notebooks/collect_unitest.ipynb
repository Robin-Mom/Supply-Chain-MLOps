```python
import pandas as pd
from pathlib import Path
import csv

def get_avis(id):
    avisF = Path(f"../data/avis_{id}k.csv")
    if avisF.exists():
        return pd.read_csv(avisF,sep=";")

df=pd.concat([get_avis(40),get_avis(28)])

nodupkeys=["auteur","note","titre","commentaire","date_experience","texteServiceReply","dateServiceReply","dataServiceReply"]
df = df.drop(columns=["url"])
df = df.sort_values(by=nodupkeys)
df = df.drop_duplicates(subset=nodupkeys)
display(df.head())

df.info(memory_usage="deep")
avis45=Path("../data/avis_45k.csv")
df.to_csv(avis45, index=None,sep=";",quoting=csv.QUOTE_ALL)
```

```python
import pandas as pd
from pathlib import Path
import csv

avis45=Path("../data/avis_45k.csv")
df45=pd.read_csv(avis45,sep=";",quoting=csv.QUOTE_ALL)


df45['date_experience']  = pd.to_datetime(df45['date_experience'], errors='coerce')
df45['date_avis']        = pd.to_datetime(df45['date_avis'], errors='coerce')
df45['dateServiceReply'] = pd.to_datetime(df45['dateServiceReply'], errors='coerce')

df45['Period']= 'P'+df45['date_experience'].dt.isocalendar().year.astype(str) + \
     df45['date_experience'].dt.isocalendar().week.astype(str).str.zfill(2)

df45['YearMonth_experience']  = df45['date_experience'].dt.to_period('M')
df45['YearMonth_avis']        = df45['date_avis'].dt.to_period('M')
df45['YearMonthServiceReply'] = df45['dateServiceReply'].dt.to_period('M')


df45
```
